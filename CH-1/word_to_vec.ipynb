```python
import nltk
import gensim.downloader as api
from nltk.tokenize import sent_tokenize,word_tokenize
from gensim.models import Word2Vec
```

```python
nltk.download("all")
```

```python
data = api.load("text8")
```

```python
with open("data.txt", "r", encoding="utf-8") as f:
    c = f.read()
    print(c)
c
```

```python
sentances = sent_tokenize(c)
# Filter out sentences with no words (e.g., empty or whitespace-only)
filtered_sentences = [x.strip() for x in sentances if x.strip()]
```

```python
with open("x.txt", "w", encoding="utf-8") as f:
    for sentence in filtered_sentences:
        f.write(sentence + "\n")

```

```python
import string

# Create vocabulary for each sentence
sentences_vocab = []

for sentence in filtered_sentences:
    # Tokenize sentence into words
    words = word_tokenize(sentence.lower())
    
    # Remove punctuation
    words = [word for word in words if word not in string.punctuation]
    
    sentences_vocab.append(words)
```

```python
sentences_vocab
```

```python
w2v_cbow = Word2Vec(sentences_vocab, vector_size=100, window=5, min_count=1, sg=0)

w2v_sg = Word2Vec(sentences_vocab, vector_size=100, window=5, min_count=1, sg=1)

```
