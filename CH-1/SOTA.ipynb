```python
# ==================== EMBEDDING FUNCTIONS ====================

# 1. GOOGLE GENERATIVE AI (Gemini)
def get_embedding_gemini(text):
    """Get embeddings using Google's Gemini model"""
    if not gemini_api_key:
        raise ValueError("GEMINI_TOKEN not found in .env file")
    
    genai.configure(api_key=gemini_api_key)
    result = genai.embed_content(
        model='models/gemini-embedding-001',
        content=text,
        task_type='retrieval_document'
    )
    return result['embedding']


# 2. OPENAI (Text Embedding Model)
def get_embedding_openai(text):
    """Get embeddings using OpenAI's embedding model"""
    if not openai_api_key:
        raise ValueError("OPENAI_API_KEY not found in .env file")
    
    client = OpenAI(api_key=openai_api_key)
    response = client.embeddings.create(
        model="text-embedding-3-small",  # or "text-embedding-3-large" for better quality
        input=text
    )
    return response.data[0].embedding


# 3. HUGGING FACE (Sentence Transformers - Local)
def get_embedding_huggingface(text, model_name='sentence-transformers/all-MiniLM-L6-v2'):
    """Get embeddings using Hugging Face Sentence Transformers (runs locally)"""
    model = SentenceTransformer(model_name)
    embedding = model.encode(text)
    return embedding.tolist()


# ==================== WRAPPER FUNCTION ====================
def get_embedding(text, provider='huggingface'):
    """
    Get embeddings using the specified provider
    
    Parameters:
    - text: The input text to embed
    - provider: 'gemini', 'openai', or 'huggingface' (default)
    """
    if provider.lower() == 'gemini':
        return get_embedding_gemini(text)
    elif provider.lower() == 'openai':
        return get_embedding_openai(text)
    elif provider.lower() == 'huggingface':
        return get_embedding_huggingface(text)
    else:
        raise ValueError(f"Unknown provider: {provider}. Use 'gemini', 'openai', or 'huggingface'")


# ==================== EXAMPLE USAGE ====================
sentence = "This is a sample sentence."
paragraph = "This is a sample paragraph. It contains multiple sentences and provides more context for embedding."

# Try with Hugging Face (default - no API key needed)
sentence_embedding = get_embedding(sentence, provider='huggingface')
paragraph_embedding = get_embedding(paragraph, provider='huggingface')

print(f"Sentence embedding length (Hugging Face): {len(sentence_embedding)}")
print(f"Paragraph embedding length (Hugging Face): {len(paragraph_embedding)}")
```

```python
# ==================== COMPARISON: SIMILARITY BETWEEN EMBEDDINGS ====================
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Compare embeddings from different providers
hf_embeddings = [
    get_embedding(sentence, provider='huggingface'),
    get_embedding(paragraph, provider='huggingface')
]

# Calculate similarity between sentence and paragraph
similarity = cosine_similarity([hf_embeddings[0]], [hf_embeddings[1]])[0][0]
print(f"Cosine similarity (Sentence vs Paragraph): {similarity:.4f}")
```
