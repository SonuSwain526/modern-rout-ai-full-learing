onehot ecoding

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
sentances = ["apple is good", "apple is fruit", "fruit is good"]
token = [[x] for x in sentances]
```

```python
word = [wordd.split() for wordd in sentances]
x = [[z] for w in word for z in w]
x
```

```python
encoder = OneHotEncoder()
encoder.fit(x)
```

```python
encoder.categories_[0]
```

```python
result = encoder.transform([["apple"], ["is"]]).toarray()
result
```

bag of words (bow)

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
bow = CountVectorizer()
```

```python
bow.fit_transform(sentances).toarray()
```

```python
bow.transform(["apple is apple"]).toarray()
```
