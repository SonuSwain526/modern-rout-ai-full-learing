```python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage

# 1. Load the variables from your .env file into the system environment
load_dotenv()
```

```python

# 2. Initialize the model. 
# LangChain automatically looks for the "OPENAI_API_KEY" variable behind the scenes.
llm = ChatOpenAI(
    model="gpt-5.5",
    temperature=0.7
)
```

```python

# 3. Format your prompt instructions
messages = [
    SystemMessage(content="You are a helpful and precise AI coding assistant."),
    HumanMessage(content="how many staed are in india and give me top 5 dirties state of india")
]


# 4. Run the model
print("Sending request to OpenAI...")
response = llm.invoke(messages)

# 5. Print the output
print("\nResponse from GPT:")
print(response.content)
```

```python
from dotenv import load_dotenv
from langchain_groq import ChatGroq
from langchain_core.messages import HumanMessage, SystemMessage
from langgraph.graph import StateGraph, START, END, MessagesState
```

```python
load_dotenv(override=True)
```

```python
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0.3
)
```

```python
# Step A: Define the graph node function
def call_groq_agent(state: MessagesState):
    # Take the message history from the state and send it to Groq
    response = llm.invoke(state["messages"])
    
    # Return the new message to append it to the graph state
    return {"messages": [response]}
```

```python
builder = StateGraph(MessagesState)
```

```python
builder.add_node("groq_node", call_groq_agent)

builder.add_edge(START, "groq_node")
builder.add_edge("groq_node", END)
```

```python
app = builder.compile()
```

```python
inputs = {
    "messages": [
        SystemMessage(content="You are an ultra-fast AI helper powered by Groq LPUs."),
        HumanMessage(content="Explain the difference between compiler and interpreter in one short bulleted list.")
    ]
}

print("Streaming response through LangGraph utilizing Groq LPUs...\n")
for output in app.stream(inputs, stream_mode="values"):
    last_message = output["messages"][-1]
    print(f"[{last_message.type.upper()}]: {last_message.content}\n")
```
